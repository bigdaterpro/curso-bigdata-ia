### La pregunta buena: ¿cómo cabe un modelo de 8.000 millones de parámetros en un portátil?

En crudo pesarían **~16 GB** — imposible en RAM de consumo. La respuesta es la **cuantización**:
guardar cada peso con **menos precisión** (4 bits en vez de 16). El modelo pierde un poco de finura
y pasa a **~4,7 GB**. Cabe.

El precio restante lo pone la CPU: **generar un token exige recorrer toda la red**, y sin GPU eso
son pocas palabras por segundo.

**El MISMO prompt, cinco ejecuciones.** Regenera cinco veces y pega las cinco respuestas abajo.

````python
import json

CLAVES = {"id", "nombre", "categoria", "riesgo_stock"}

respuestas = [
    """[pega aquí la 1]""",
    """[pega aquí la 2]""",
    """[pega aquí la 3]""",
    """[pega aquí la 4]""",
    """[pega aquí la 5]""",
]

validas = 0
for i, r in enumerate(respuestas, 1):
    try:
        d = json.loads(r)
        ok = len(d) == 10 and all(set(o.keys()) == CLAVES for o in d)
        marca = "OK    " if ok else "RARO  "
        claves = "exactas" if ok else "distintas"
        print(f"  {i}. {marca} {len(d)} objetos · claves {claves}")
        validas += ok
    except Exception as e:
        print(f"  {i}. ROTO   {type(e).__name__}: {str(e)[:60]}")

print()
print(f"  ESTABILIDAD: {validas} de 5")
````

# LAB14 · La API con tu clave: de `curl` al bucle
### ◉ Aquí dentro. Y aquí se corona el bloque.

Todo lo que has hecho en el navegador existe como **endpoint HTTP**: mandas un POST con JSON y
vuelve un JSON. Tres conceptos y una liturgia:

- **La clave de API es tu identidad y tu contador.** Se trata **como una contraseña**: variable de
  entorno, **jamás en un repositorio, jamás en un cuaderno entregado**. Si se escapa, se revoca en
  dos clics y se genera otra.

- **Los tokens** del Bloque 1 son aquí la **unidad de facturación** (entrada + salida).
- **El error `429` no es una avería: es cuota.** Se espera un minuto. El `403` sí es la clave.

> ## ⭐ EL PATRÓN QUE CORONA EL CURSO
>
> ### **la IA GENERA (el SQL) → tu motor EJECUTA (DuckDB) → tu ancla VERIFICA (el número)**

## Paso 0 · La clave, con su liturgia · `BASE`

En la ▸ **Terminal de Jupyter**:

````python
import os

API_KEY = os.environ.get("GEMINI_API_KEY", "")
# API_KEY = "pega-aqui-tu-clave-SOLO-para-la-clase"   # <-- y BÓRRALA antes de entregar

print("Clave cargada." if API_KEY
      else "  FALTA la clave: export GEMINI_API_KEY=... y reinicia el kernel")
````

---

## Paso 1 · La primera llamada · `BASE`

**Mírala por dentro antes de ejecutarla.** No hay magia: es un POST con JSON, la temperatura que
tocaste en el taller, y la respuesta desempaquetada de su envoltorio.

````python
import json, urllib.request, urllib.error

MODELO = "gemini-2.0-flash"

def gemini(prompt, temperatura=0.2):
    url = (f"https://generativelanguage.googleapis.com/v1beta/models/"
           f"{MODELO}:generateContent?key={API_KEY}")
    cuerpo = {"contents": [{"parts": [{"text": prompt}]}],
              "generationConfig": {"temperature": temperatura}}
    peticion = urllib.request.Request(
        url, data=json.dumps(cuerpo).encode("utf-8"),
        headers={"Content-Type": "application/json"}, method="POST")
    try:
        with urllib.request.urlopen(peticion) as r:
            datos = json.load(r)
        return datos["candidates"][0]["content"]["parts"][0]["text"]
    except urllib.error.HTTPError as e:
        if e.code == 429:
            return "[429] Cuota del nivel gratuito. No es avería: espera un minuto."
        if e.code in (400, 403):
            return ("[403] La clave no ha cargado."
                    " Repite el paso 0 y reinicia el kernel.")
        return f"[{e.code}] {e.reason}"

print(gemini("Responde solo con una palabra: ¿capital de Aragón?"))

# Criterio: responde con sentido. La forma exacta puede variar -- no es determinista,
# y ya sabes qué hacer con eso.
````

## Paso 2 · El patrón, una vez · `BASE`

Le damos **el esquema de tu Parquet como contexto**, le pedimos **solo** la consulta, la limpiamos
de adornos, la ejecutamos con DuckDB… **y la juzga el ancla.**

````python
import duckdb

def limpiar_sql(texto):
    """La IA envuelve el SQL en adornos. Higiene antes de ejecutar."""
    t = texto.strip()
    if t.startswith("```"):
        t = t.split("```")[1]
        if t.lower().startswith("sql"):
            t = t[3:]
    return t.strip().rstrip(";").strip()

ESQUEMA = duckdb.sql(f"DESCRIBE SELECT * FROM '{RUTA}'").df().to_string(index=False)

CONTEXTO = f"""Eres un analista de datos senior.
Tengo una tabla en DuckDB accesible con esta ruta literal: '{RUTA}'
Su esquema es:
{ESQUEMA}
Responde SOLO con la consulta SQL, sin explicaciones y sin markdown."""

TAREA = ("\nTarea: la facturación total (unidades por precio unitario),"
         " redondeada a dos decimales.")
sql = limpiar_sql(gemini(CONTEXTO + TAREA))
print(sql)
print()

resultado = duckdb.sql(sql).fetchone()[0]
ANCLA = 429892547.06
print(f"  resultado: {resultado}")
print(f"  ancla:     {ANCLA}")
print("  VERIFICADO" if abs(float(resultado) - ANCLA) < 0.01
      else "  NO CUADRA -- y entonces: a auditar el SQL")
````

## ⭐ Paso 3 · El bucle completo · `COMPLETA`
### Es el ejercicio del día y el ensayo general del proyecto

Tres peticiones **en lenguaje natural** contra un diccionario de anclas: genera, ejecuta, verifica y
marca una a una.

> **Criterio: 3 de 3 verificados.**
>
> ⚠️ **Y la parte más valiosa: si alguna sale ✘, NO la borres.** Lee el SQL generado buscando **la
> decisión que el modelo tomó por ti** — ¿olvidó el `ROUND`? ¿contó los `NULL`? ¿inventó una
> columna? — reformula la petición y repite.
>
> **Esa auditoría documentada ES el ejercicio, no un contratiempo.** Es el protocolo de ayer,
> automatizado.

````python
import duckdb

ANCLAS = {
    "el número total de filas":
        999535,
    "el ticket medio: la suma de unidades por precio unitario"
    " dividida entre el número de filas, redondeada a 2 decimales":
        430.09,
    "el número de ciudades distintas (COUNT(DISTINCT) ya ignora los nulos)":
        10,
}

verificados = 0
generados = {}

for peticion, ancla in ANCLAS.items():
    sql = limpiar_sql(gemini(f"{CONTEXTO}\nTarea: {peticion}."))
    generados[peticion] = sql
    try:
        obtenido = float(duckdb.sql(sql).fetchone()[0])
        ok = abs(obtenido - float(ancla)) < 0.01
    except Exception as e:
        obtenido, ok = f"error: {type(e).__name__}", False
    verificados += bool(ok)
    marca = "OK" if ok else "**"
    print(f"  [{marca}]  {peticion[:52]:<52}  {obtenido}  (ancla {ancla})")

print()
print(f"  {verificados} de {len(ANCLAS)} verificados")
````

````python
# Si alguna falló, míralo aquí: el SQL que generó, tal cual.
if not globals().get("generados"):
    print("  Todavía no hay nada que mirar: ejecuta antes la celda del bucle.")
    print("  (Si el bucle se cortó por un 429, espera un minuto y repítelo.)")
else:
    for peticion, sql in generados.items():
        print("=" * 70)
        print(peticion)
        print("-" * 70)
        print(sql)
        print()
````

````python
import shutil, os, glob, json

SESION = 9                      # numeracion del MANUAL (sesiones 4, 5 y 6)

# ══════════════════════════════════════════════════════════════════════════
#  GUARDIÁN · ¿está en el disco lo que ves en pantalla?
#
#  Esta celda copia el FICHERO DEL DISCO, no lo que tienes delante. Jupyter
#  guarda solo cada pocos minutos: si archivas antes de un Ctrl+S, entregas
#  el cuaderno SIN tus resultados y el HTML sale vacío.
# ══════════════════════════════════════════════════════════════════════════

def resultados_en_disco(ruta):
    try:
        nb = json.load(open(ruta, encoding="utf-8"))
    except Exception:
        return 0, 0
    codigo = [c for c in nb["cells"] if c["cell_type"] == "code"]
    return sum(1 for c in codigo if c.get("outputs")), len(codigo)

cuadernos = [f for f in glob.glob("*lab11*.ipynb") if ".ipynb_checkpoints" not in f]
listo = bool(cuadernos)

if not cuadernos:
    print("  No encuentro los cuadernos de esta sesión en esta carpeta.")

for cuaderno in cuadernos:
    hechas, total = resultados_en_disco(cuaderno)
    print(f"  {cuaderno}: {hechas} de {total} celdas con resultados en el disco")
    if hechas == 0:
        listo = False

if not listo:
    print()
    print("  " + "=" * 68)
    print("   PARA AQUI. No he archivado nada.")
    print("   Pulsa  Ctrl+S  y vuelve a ejecutar ESTA celda.")
    print("  " + "=" * 68)
else:
    os.makedirs("entregables", exist_ok=True)
    piezas = cuadernos + [f for f in ["mi_bitacora.ipynb"] if os.path.exists(f)]
    for pieza in piezas:
        destino = f"entregables/S{SESION:02d}_{os.path.basename(pieza)}"
        shutil.copy(pieza, destino)
        print("  copiado:", destino)
````

````python
import glob, os, re

# ══════════════════════════════════════════════════════════════════════════
#  LA COMPROBACIÓN QUE CIERRA EL CÍRCULO
#
#  Un cuaderno ejecutado deja en el HTML el número de cada celda: [1]:, [2]:...
#  Si no hay ninguno, el HTML NO lleva tus resultados.
#  El TAMAÑO del fichero engaña; este número, no.
# ══════════════════════════════════════════════════════════════════════════

def celdas_ejecutadas(ruta_html):
    h = open(ruta_html, encoding="utf-8", errors="ignore").read()
    return len(re.findall(r"\[[0-9]+\]:", h)), h.count("data:image/png;base64")

htmls = sorted(glob.glob("entregables/*.html"))
vacios = []

if not htmls:
    print("  No hay ningún HTML. ¿Ejecutaste la celda anterior?")

for h in htmls:
    ejecutadas, graficas = celdas_ejecutadas(h)
    kb = os.path.getsize(h) / 1024
    if ejecutadas == 0:
        vacios.append(os.path.basename(h))
    estado = "OK    " if ejecutadas else "VACIO "
    print(f"  {estado} {os.path.basename(h):<34}"
          f" {ejecutadas:>3} celdas ejecutadas · {kb:.0f} KB")

print()
if vacios:
    print("  Estos HTML no llevan resultados:", ", ".join(vacios))
    print("  Ctrl+S y repite las dos celdas anteriores.")
elif htmls:
    print("  Todo correcto. Ya puedes empaquetar.")
````
