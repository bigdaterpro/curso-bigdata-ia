# FASE 1 · EL PIPELINE · **40 puntos**

Revalida el camino entero: datasets generados → tu ETL de Spark → **el Parquet maestro con sus
anclas de entrada**.

**Si todo vive de las sesiones anteriores, esta fase son quince minutos.** Si algo se perdió, aquí
demuestras que tu proceso es **reproducible** — que era, literalmente, el encargo.

## La celda-ancla

**Se juzga sola.** Ejecuta las tres comprobaciones y grita `VALIDADO`… **o se niega a seguir**: el
`assert` detiene el cuaderno.

> 🗣️ **Conócelo antes de que te pare él a ti.**

````python
import duckdb

RUTA = '../datasets/salida/ventas_limpio.parquet/*.parquet'
#      ^ ojo al comodín: Spark escribe una CARPETA, no un fichero

ANCLAS = {"filas": 999535, "ticket": 430.09, "ciudades": 10}

try:
    filas = duckdb.sql(f"SELECT COUNT(*) FROM '{RUTA}'").fetchone()[0]
    ticket = duckdb.sql(f"""SELECT ROUND(SUM(unidades*precio_unitario)/COUNT(*), 2)
                            FROM '{RUTA}'""").fetchone()[0]
    ciudades = duckdb.sql(f"""SELECT COUNT(DISTINCT ciudad) FROM '{RUTA}'
                              WHERE ciudad IS NOT NULL""").fetchone()[0]
except Exception as e:
    print("  No he podido leer el Parquet maestro.")
    print(" ", e)
    print()
    print("  No es un castigo: vuelve a ejecutar tu cuaderno del LAB10 y regenéralo.")
    print("  ESO es demostrar que el pipeline es reproducible, que era el encargo.")
    raise SystemExit

print(f"  filas    {filas}      (esperado {ANCLAS['filas']})")
print(f"  ticket   {ticket}     (esperado {ANCLAS['ticket']})")
print(f"  ciudades {ciudades}          (esperado {ANCLAS['ciudades']})")
print()

assert filas == ANCLAS["filas"],           f"filas: {filas} != {ANCLAS['filas']}"
assert abs(float(ticket) - ANCLAS["ticket"]) < 0.01, \
    f"ticket: {ticket} != {ANCLAS['ticket']}"
assert ciudades == ANCLAS["ciudades"], \
    f"ciudades: {ciudades} != {ANCLAS['ciudades']}"

print("  PIPELINE VALIDADO  --  Fase 1 superada")
````

## ② La comprobación de gobierno — **antes de exportar nada**

**−10 puntos si viaja una clave.** Esta celda la busca por ti.

````python
import glob, json, re

sospechosos = []
for ruta in glob.glob("*proyecto*.ipynb"):
    if ".ipynb_checkpoints" in ruta:
        continue
    nb = json.load(open(ruta, encoding="utf-8"))
    for celda in nb["cells"]:
        for linea in "".join(celda["source"]).split("\n"):
            if linea.strip().startswith("#"):
                continue
            if re.search(r"AIza[0-9A-Za-z_\-]{20,}", linea):
                sospechosos.append((ruta, linea.strip()[:60]))

if sospechosos:
    print("  " + "=" * 66)
    print("   PARA. Hay algo que parece una clave dentro del cuaderno:")
    for ruta, linea in sospechosos:
        print(f"     {ruta}:  {linea}")
    print("   Bórrala, Ctrl+S, y vuelve a ejecutar ESTA celda.")
    print("  " + "=" * 66)
else:
    print("  Sin claves a la vista.")

# Mira el fichero EN DISCO: si acabas de borrarla, pulsa Ctrl+S antes.
````
