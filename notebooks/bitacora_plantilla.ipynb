## 5 · Dos formas de hablar con la máquina

Hasta ahora has usado `!`, que significa *«esto no es Python: pásaselo a la terminal»*.
Cuando quieras mandar **varias** órdenes seguidas, hay una forma más cómoda:


````python
%%bash
# Todo lo que hay en esta celda es terminal, no Python.
echo "Ficheros disponibles:"
ls ../datasets/ | wc -l
echo "Espacio que ocupan en total:"
du -sh ../datasets/
````

Y esta celda, en cambio, es Python de verdad — el mismo lenguaje que usarás para todo lo demás:


````python
import sys, platform

print("Python  :", sys.version.split()[0])
print("Sistema :", platform.system(), platform.machine())
print()
print("Ventas registradas:", f"{1_000_000:,}".replace(",", "."))
````

> **Un cuaderno, dos idiomas.** La celda de arriba es Python; las anteriores eran terminal.
> Puedes mezclarlos en el mismo documento, y eso es justo lo que hace útil a esta herramienta:
> el análisis, las órdenes del sistema y tus notas viven en el mismo sitio y se leen en orden.

Fíjate también en `platform.machine()`: dice **`x86_64`**. Esa palabra es la que te va a permitir
montar un sistema de almacenamiento distribuido en la sesión 3.
